```python
def reset(day):
    if day==8:
        return 1
    elif day==9:
        return 2
    return day
```

```python
def countSundays(day):
    if day==1 or day==2 or day==3 or day==4:
        return 4
    else:
        return 5
```

```python
sumOfSundays=0
```

```python
day=2 #tuesday
for year in range(1901,2001):
    day=reset(day)

    sumOfSundays+=countSundays(day)

    if year%100==0:
        if year%400==0:
            day+=2
        else:
            day+=1
    elif year%4==0:
        day+=2
    else:
        day+=1
    
    
```
