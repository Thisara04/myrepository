```python
#get the product of adjacent numbers in the list
#changing numOfAdjacents change how many numbers are multiplied
def multiDigit(ind,digit,numOfAdjacents):
    product=1
    #ex.- get the production from index 3 to 3+13 of digit list
    for i in range(ind,ind+numOfAdjacents):
        product=product*digit[i]
    return product
```
