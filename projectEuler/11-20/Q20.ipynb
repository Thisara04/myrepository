```python
def factorial(num):
    if num==1:
        return 1
    else:
        return num*factorial(num-1)
```

```python
def factDigitSum(num):
    factNum=factorial(num)
    strNum=str(factNum)
    sum=0
    for digit in strNum:
        sum+=int(digit)
    return sum
```

```python
print(factDigitSum(100))
```
