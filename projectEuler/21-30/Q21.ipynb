```python
#output the divisor list of number "n" excluding "n"
def divisors(num):
    divList=[]
    for i in range(1,int(num**0.5)+1):
        if num%i==0:
            if i==num/i:
                divList.append(i)
            else:
                divList.append(i)
                divList.append(int(num/i))
                
    if num in divList:
        divList.remove(num)
    return divList
```

```python
#give the sum of the divisors
#input a list output an int
def d(n):
    divSum=sum(divisors(n))
    return divSum
```

```python
amicSum=0

for a in range(10000):
    
    #check for amicables
    if a==d(d(a)) and a!=d(a):
        amicSum+=a
```
